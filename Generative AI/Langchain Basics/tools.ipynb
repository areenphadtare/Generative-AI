```python
from langchain_core.tools import tool
@tool
def division(a: float, b: float) -> float:
    """Divides two numbers."""
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b    
```

```python
## async implementation
from langchain_core.tools import tool
@tool
async def async_division(a: float, b: float) -> float:
    """Divides two numbers asynchronously."""
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b
    
```

```python
from typing import Annotated,List
@tool
def multiply_by_max(a:Annotated[int,"A value"],b:Annotated[int,"B value"])->int:
    """Multiplies two numbers by the maximum of the two."""
    max_value = max(a, b)
    return a * max_value, b * max_value
```

### Structured Tool
The structured Tool from _function class method provides a bit more configuration than the @tool_decorator,without requesting much additional code.

```python
from langchain_core.tools import StructuredTool
def multiply(a: int, b: int) -> int:
    """Multiplies two numbers."""
    return a * b

async def async_multiply(a: int, b: int) -> int:
    """Multiplies two numbers asynchronously."""
    return a * b    
calculator = StructuredTool.from_function(func=multiply,coroutine=async_multiply)

print (calculator.invoke({"a": 5, "b": 10}))
print(await calculator.ainvoke({"a": 5, "b": 10}))
```
